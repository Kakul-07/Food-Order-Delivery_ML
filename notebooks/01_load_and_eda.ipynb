# 01 — Load and explore datasets
Inspect file paths, shape, data types, missing values, duplicates, and sample rows before changing anything.

```python
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
RAW = ROOT/"data/raw"
PROCESSED = ROOT/"data/processed"
REPORTS = ROOT/"reports"
PROCESSED.mkdir(parents=True, exist_ok=True)
REPORTS.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
for p in sorted(RAW.glob("*")):
    print(p.name, f"{p.stat().st_size:,} bytes")
```

```python
files = {
    "delivery_train": "delivery_time_train.csv",
    "delivery_test": "delivery_time_test.csv",
    "online_orders": "online_food_orders.csv",
    "restaurants": "swiggy_restaurant_listings.csv",
}
data = {}
for key, name in files.items():
    p = RAW/name
    if p.exists():
        data[key] = pd.read_csv(p, low_memory=False, encoding="latin1" if key == "restaurants" else "utf-8")
        print(f"{key}: {data[key].shape}")
    else:
        print("MISSING:", p)
```

```python
for name, df in data.items():
    print("\n---", name, "---")
    display(df.head(5))
    display(pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "missing": df.isna().sum(),
        "unique": df.nunique(dropna=True)
    }).sort_values("missing", ascending=False))
    print("Exact duplicate rows:", int(df.duplicated().sum()))
```

```python
# Delivery target examples: strings often look like '(min) 24'.
if "delivery_train" in data:
    target_col = "Time_taken(min)"
    print(data["delivery_train"][target_col].astype(str).value_counts().head(10))
```

```python
inventory = []
for name, df in data.items():
    inventory.append({
        "dataset": name, "rows": len(df), "columns": len(df.columns),
        "duplicate_rows": int(df.duplicated().sum()),
        "missing_cells": int(df.isna().sum().sum())
    })
inventory = pd.DataFrame(inventory)
display(inventory)
inventory.to_csv(REPORTS/"data_inventory.csv", index=False)
```
